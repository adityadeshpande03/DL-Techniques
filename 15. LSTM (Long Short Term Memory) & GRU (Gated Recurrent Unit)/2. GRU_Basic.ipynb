```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
```

```python
tokenizer.word_index
```

```python
input_sequences = []
for sentence in text.split('\n'):
    # Skip empty sentences
    if not sentence.strip():
        continue

    # Convert sentence to sequence
    tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

    # Create subsequences
    for i in range(1, len(tokenized_sentence)):
        input_sequences.append(tokenized_sentence[:i+1])
```

```python
X = padded_sequences[:,:-1]
y = padded_sequences[:,-1]
```

```python
vocab_size = len(tokenizer.word_index) + 1  # Add 1 for padding token
```

```python
y_one_hot = to_categorical(y, num_classes=vocab_size)
```

```python
print("Input shape (X):", X.shape)
print("Output shape (y_one_hot):", y_one_hot.shape)
print("Vocabulary size:", vocab_size)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Embedding,GRU
```

```python
model = Sequential()

model.add(Embedding(360,100,input_length=17))
model.add(GRU(150))
model.add(Dense(360,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y_one_hot,epochs=100)
```

```python
import numpy as np

import time
text_new = "She smiled"

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text_new])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text_new = text_new + " " + word
      print(text_new)
      time.sleep(2)
```
